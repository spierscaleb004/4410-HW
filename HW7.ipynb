```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import time

def build_convent():
  model = keras.Sequential([
      layers.Conv2D(32, (3, 3), activation='relu', input_shape=(28, 28, 1)),
      layers.MaxPooling2D((2, 2)),
      layers.Conv2D(64, (3, 3), activation='relu'),
      layers.MaxPooling2D((2, 2)),
      layers.Conv2D(64, (3, 3), activation='relu'),
      layers.Flatten(),
      layers.Dense(64, activation='relu'),
      layers.Dense(10, activation='softmax')
  ])
  model.compile(optimizer='adam',
                loss='sparse_categorical_crossentropy',
                metrics=['accuracy'])
  return model

# Loading the datasets
(x_train_mnist, y_train_mnist), (x_test_mnist, y_test_mnist) = keras.datasets.mnist.load_data()
(x_train_fashion, y_train_fashion), (x_test_fashion, y_test_fashion) = keras.datasets.fashion_mnist.load_data()

x_train_mnist = x_train_mnist.reshape(-1, 28, 28, 1).astype("float32") / 255
x_test_mnist  = x_test_mnist.reshape(-1, 28, 28, 1).astype("float32") / 255
x_train_fashion = x_train_fashion.reshape(-1, 28, 28, 1).astype("float32") / 255
x_test_fashion  = x_test_fashion.reshape(-1, 28, 28, 1).astype("float32") / 255

model_mnist = build_convent()
start = time.time()
history_mnist = model_mnist.fit(x_train_mnist, y_train_mnist, epochs=5, batch_size=64, validation_split=0.1, verbose=2)
mnist_time = time.time() - start
test_loss_mnist, test_acc_mnist = model_mnist.evaluate(x_test_mnist, y_test_mnist, verbose=0)

model_fashion = build_convent()
start = time.time()
history_fashion = model_fashion.fit(x_train_fashion, y_train_fashion, epochs=5, batch_size=64, validation_split=0.1, verbose=2)
fashion_time = time.time() - start
test_loss_fashion, test_acc_fashion = model_fashion.evaluate(x_test_fashion, y_test_fashion, verbose=0)

print("MNIST Test Accuracy:", test_acc_mnist, " Training Time:", mnist_time)
print("Fashion-MNIST Test Accuracy:", test_acc_fashion, " Training Time:", fashion_time)

def build_convnet_no_dense():
    model = keras.Sequential([
        layers.Conv2D(32, (3, 3), activation="relu", input_shape=(28, 28, 1)),
        layers.MaxPooling2D((2, 2)),
        layers.Conv2D(64, (3, 3), activation="relu"),
        layers.MaxPooling2D((2, 2)),
        layers.Conv2D(64, (3, 3), activation="relu"),
        layers.Flatten(),
        layers.Dense(10, activation="softmax")
    ])
    model.compile(optimizer="adam",
                  loss="sparse_categorical_crossentropy",
                  metrics=["accuracy"])
    return model

model_no_dense = build_convnet_no_dense()
history_no_dense = model_no_dense.fit(x_train_fashion, y_train_fashion, epochs=5, batch_size=64, validation_split=0.1, verbose=2)
test_loss_no_dense, test_acc_no_dense = model_no_dense.evaluate(x_test_fashion, y_test_fashion, verbose=0)

print("Fashion-MNIST Test Accuracy (no Dense layer):", test_acc_no_dense)

def build_convnet_big_dense():
    model = keras.Sequential([
        layers.Conv2D(32, (3, 3), activation="relu", input_shape=(28, 28, 1)),
        layers.MaxPooling2D((2, 2)),
        layers.Conv2D(64, (3, 3), activation="relu"),
        layers.MaxPooling2D((2, 2)),
        layers.Conv2D(64, (3, 3), activation="relu"),
        layers.Flatten(),
        layers.Dense(4096, activation="relu"),  # Very large layer
        layers.Dense(64, activation="relu"),
        layers.Dense(10, activation="softmax")
    ])
    model.compile(optimizer="adam",
                  loss="sparse_categorical_crossentropy",
                  metrics=["accuracy"])
    return model

model_big_dense = build_convnet_big_dense()
history_big_dense = model_big_dense.fit(x_train_fashion, y_train_fashion, epochs=5, batch_size=64, validation_split=0.1, verbose=2)
test_loss_big_dense, test_acc_big_dense = model_big_dense.evaluate(x_test_fashion, y_test_fashion, verbose=0)

print("Fashion-MNIST Test Accuracy (with Dense 4096):", test_acc_big_dense)
```
